```python
import pandas as pd
import matplotlib.pyplot as plt

# Dados de estimativa semanal de uso do celular (em minutos por dia)
dados_uso = {
    'Dia': ['Seg', 'Ter', 'Qua', 'Qui', 'Sex', 'Sáb', 'Dom'],
    'Redes Sociais': [120, 90, 110, 105, 140, 210, 180],
    'Estudos/Pesquisa': [60, 80, 75, 90, 45, 20, 15],
    'Jogos/Lazer': [45, 30, 40, 50, 90, 150, 120]
}

# Criando a tabela de dados (DataFrame)
df = pd.DataFrame(dados_uso)
df['Tempo Total (min)'] = df['Redes Sociais'] + df['Estudos/Pesquisa'] + df['Jogos/Lazer']
df['Tempo Total (horas)'] = (df['Tempo Total (min)'] / 60).round(1)

print("--- RELATÓRIO SEMANAL DE USO ---")
print(df[['Dia', 'Redes Sociais', 'Estudos/Pesquisa', 'Jogos/Lazer', 'Tempo Total (horas)']])


```

```python
# Criando o gráfico de colunas empilhadas
plt.figure(figsize=(9, 5))
plt.bar(df['Dia'], df['Redes Sociais'], label='Redes Sociais', color='#4A90E2')
plt.bar(df['Dia'], df['Estudos/Pesquisa'], bottom=df['Redes Sociais'], label='Estudos', color='#50E3C2')
plt.bar(df['Dia'], df['Jogos/Lazer'], bottom=df['Redes Sociais'] + df['Estudos/Pesquisa'], label='Jogos/Lazer', color='#F5A623')

plt.title('Distribuição do Tempo de Tela Semanal (Minutos)')
plt.xlabel('Dia da Semana')
plt.ylabel('Minutos de Uso')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
